```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Importing necessary libraries for data analysis and visualization.

### 🟢3. Load Dataset

```python
df= pd.read_csv("mental_health_dataset.csv")
df.head()
```

Loading dataset and displaying first 5 rows.

### 🟢5. Data Cleaning

```python
df=df.dropna()
```

Removing missing values to ensure clean analysis.

### 🟢6. Exploratory Data Analysis (EDA)

#### 📊6.1. Histogram (Stress Distribution)

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Stress_Level'], kde=True)
plt.title("Stress Level Distribution")
plt.savefig("stress_boxplot.png",dpi=150,bbox_inches='tight')
plt.show()
plt.close()
```

Most students fall under moderate stress levels, but a significant portion experiences high stress.

#### 📊6.2. Counterplot (Gender Comparison)

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.savefig("gender_distribution.png",dpi=150,bbox_inches='tight')
plt.show()
plt.close()
```

The dataset shows a balanced/slightly skewed gender distribution

#### 📊6.3. Scatter Plot (Sleep vs Stress)

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Sleep_Hours', y='Stress_Level', data=df)
plt.title("Sleep vs Stress")
plt.savefig("sleep_vs_stress.png",dpi=150,bbox_inches='tight')
plt.show()
plt.close()
```

Lower sleep hours are associated with higher stress levels.

#### 📊6.4. Boxplot (Outliers Detection)

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Stress_Level'])
plt.title("Stress Level Boxplot")
plt.savefig("outlier_detection.png",dpi=150,bbox_inches='tight')
plt.show()
plt.close()
```

Outliers indicate extreme stress cases that may require attention.

#### 📊6.5. Corelation Heatmap 

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.savefig("corelation.png",dpi=150,bbox_inches='tight')
plt.show()
plt.close()
```

Strong correlations exist between sleep, stress, and mental health indicators.

### 📊6.6.Line chart(sleep vs stress)

```python
plt.figure(figsize=(6,4))
plt.plot(df['Sleep_Hours'],df['Stress_Level'])
plt.title("Sleep vs Stress")
plt.xlabel("Sleep_Hours")
plt.ylabel("Stress_Level")
plt.savefig("SleepvsStress(plot).png", dpi=150, bbox_inches='tight')
plt.show()
plt.close()
```

Lower sleep hours are associated with higher stress levels.

### 🟢 7. Key Findings

1. Students with lower sleep duration tend to have higher stress levels.
2. Moderate to high stress is common among students.
3. Certain factors like academic pressure strongly influence mental health.
4. Outliers indicate a group of highly stressed individuals.

### 🟢8. Conclusion 

This analysis highlights the importance of sleep and lifestyle in managing student stress levels. The findings suggest that improving sleep habits could significantly reduce stress and improve overall mental health.
